## 10. Know the difference between `bytes` and `str`

```python
a = b"h\x65llo"
print(type(a))
print(list(a))
print(a)
```

```python
a = "a\u0300 propos"
print(type(a))
print(list(a))
print(a)
```

```python
def to_str(bytes_or_str):
    if isinstance(bytes_or_str, bytes):
        value = bytes_or_str.decode('utf-8')
    else:
        value = bytes_or_str  
    return value 
```

```python
# Funkcja repr() zwraca oficjalną, tekstową reprezentację obiektu (np. stringa),
# pokazując znaki specjalne i pozwalając zobaczyć dokładną zawartość.
print(repr(to_str(b"foo")))
print(repr(to_str("bar")))
```

```python
def to_bytes(bytes_or_str):
    if isinstance(bytes_or_str, str):
        value = bytes_or_str.encode('utf-8')
    else:
        value = bytes_or_str  
    return value 
```

```python
print(repr(to_bytes(b"foo")))
print(repr(to_bytes("bar")))
```

```python
with open('text.txt', 'wb') as f:
    f.write(b'Ala ma kota')
```

```python
with open('text.txt', 'w') as f:
    f.write('Ala ma kota')
```

Obiekt "f" zwracany przez open():
- Tryb 'r' (read):
    - f.read() – odczytuje cały plik jako string
    - f.readline() – odczytuje jedną linię
    - f.readlines() – odczytuje wszystkie linie jako listę
    - iteracja po f – linia po linii
- Tryb 'w' (write):
    - f.write(text) – zapisuje tekst do pliku
    - f.writelines(lista) – zapisuje listę stringów do pliku
    - po zakończeniu pracy należy zamknąć plik: f.close() (lub użyć with)

```python
with open('text.txt', 'r') as f:
    print(f.read())
```

```python
with open('text.txt', 'rb') as f:
    print(f.read())
```

## 11. Prefer Interpolated F-strings over C-Style Format Strings and `str.format`

```python
key = 'my_var'
value = 1.234

formatted = f'{key} = {value}'
print(formatted)
```

### Wyjaśnienie kodu

`formatted = f'{key!r:<10} = {value:.2f}'`
`print(formatted)`

#### Składnia i znaczenie elementów:

* `f'...'` – **f-string** w Pythonie, umożliwia formatowanie i wstawianie wartości zmiennych bezpośrednio w tekście.
* `{key!r}` – `!r` oznacza użycie `repr(key)`, czyli reprezentacji obiektowej (np. string będzie w cudzysłowie).
* `:<10` – **wyrównanie do lewej** w polu o szerokości **10 znaków**.
* `=` – stały znak równości w środku łańcucha.
* `{value:.2f}` – liczba zmiennoprzecinkowa sformatowana z **dokładnością do 2 miejsc po przecinku**.

```python
formatted = f'{key!r:<10} = {value:.2f}'
print(formatted)
```

## 12. Understand the Difference between `repr` and `str` when printing objects

```python
print("foo bar")
```

This is equivalent to all of these alternatives:
* Calling the `str` function before passing the value to print
* Using the `"%s"` format string with the `%` operator
* Using the default formatting of the value with an f-string
* Calling the `format` built-in function
* Explicitly calling the `__format__` special method
* Explicitly calling the `__str__` special method


```python
my_value = "foo bar"
print(str(my_value))
print("%s" % my_value)
print(f"{my_value}")
print(format(my_value))
print(my_value.__format__("s"))
print(my_value.__str__())
```

```python
int_value = 5
str_value = "5"
print(int_value)
print(str_value)
print(f"Is {int_value} == {str_value}?")
```

```python
print(f'{str_value!r}') # to samo co repr
print("Is %r == %r?" % (int_value, str_value)) # tak samo %r dziala jak repr
```

```python
class OpaqueClass:
    def __init__(self, x, y):
        self.x = x
        self.y = y
        
obj = OpaqueClass(1, "foo")
print(obj)
```

```python
class BetterClass:
    def __init__(self, x, y):
        self.x = x
        self.y = y
        
    def __repr__(self):
        return f"BetterClass({self.x!r}, {self.y!r})"
```

```python
obj = BetterClass(2, "foo")
print(obj)
```

```python
print(str(obj))
```

```python
class StringifiableBetterClass(BetterClass):
    def __str__(self):
        return f"({self.x}, {self.y})"
```

```python
obj2 = StringifiableBetterClass(2, "bar")
print("Human readable:", obj2)
print("Printable: ", repr(obj2))
```

## 13: Prefer Explicit String Concatenation over Implicit, Especially in Lists


```python
my_test1 = "hello" "world"
my_test2 = "hello" + "world"
assert my_test1 == my_test2
```

```python
x = 1
my_test1 = (
r"first \ part is here with escapes\n, "
f"string interpolation {x} in here, "
'this has "double quotes" inside'
)
print(my_test1)
```

```python
y = 2
my_test2 = r"fir\st" f"{y}" '"third"'
print(my_test2)
```

```python
my_test3 = r"fir\st", f"{y}" '"third"'
print(my_test3)
```

```python
my_test4 = [
"first line\n",
"second line\n",
"third line\n",
]
print(my_test4)
```

```python
my_test5 = [
"first line\n",
"second line\n" # Comma removed
"third line\n",
]
print(my_test5)
```

```python
my_test6 = [
"first line\n",
"second line\n" + # Explicit
"third line\n",
]
assert my_test5 == my_test6
```

## 14: Know How to Slice Sequences

The basic form of the slicing syntax is `somelist[start:end]` ,
where `start` is inclusive and `end` is exclusive:

```python
a = ["a", "b", "c", "d", "e", "f", "g", "h"]
print("Middle two: ", a[3:5])
print("All but ends:", a[1:7])
```
