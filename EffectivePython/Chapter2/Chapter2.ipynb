```python
a = b"h\x65llo"
print(type(a))
print(list(a))
print(a)
```

```python
def to_str(bytes_or_str):
    if isinstance(bytes_or_str, bytes):
        value = bytes_or_str.decode('utf-8')
    else:
        value = bytes_or_str  
    return value 
```

```python
# Funkcja repr() zwraca oficjalną, tekstową reprezentację obiektu (np. stringa),
# pokazując znaki specjalne i pozwalając zobaczyć dokładną zawartość.
print(repr(to_str(b"foo")))
print(repr(to_str("bar")))
```

```python
def to_bytes(bytes_or_str):
    if isinstance(bytes_or_str, str):
        value = bytes_or_str.encode('utf-8')
    else:
        value = bytes_or_str  
    return value 
```

```python
print(repr(to_bytes(b"foo")))
print(repr(to_bytes("bar")))
```

## 11. Prefer Interpolated F-strings over C-Style Format Strings and `str.format`

```python
key = 'my_var'
value = 1.234

formatted = f'{key} = {value}'
print(formatted)
```

### Wyjaśnienie kodu

`formatted = f'{key!r:<10} = {value:.2f}'`
`print(formatted)`

#### Składnia i znaczenie elementów:

* `f'...'` – **f-string** w Pythonie, umożliwia formatowanie i wstawianie wartości zmiennych bezpośrednio w tekście.
* `{key!r}` – `!r` oznacza użycie `repr(key)`, czyli reprezentacji obiektowej (np. string będzie w cudzysłowie).
* `:<10` – **wyrównanie do lewej** w polu o szerokości **10 znaków**.
* `=` – stały znak równości w środku łańcucha.
* `{value:.2f}` – liczba zmiennoprzecinkowa sformatowana z **dokładnością do 2 miejsc po przecinku**.

```python
formatted = f'{key!r:<10} = {value:.2f}'
print(formatted)
```

## 12. Understand the Difference between `repr` and `str` when printing objects

```python
print("foo bar")
```

This is equivalent to all of these alternatives:
* Calling the `str` function before passing the value to print
* Using the `"%s"` format string with the `%` operator
* Using the default formatting of the value with an f-string
* Calling the `format` built-in function
* Explicitly calling the `__format__` special method
* Explicitly calling the `__str__` special method


```python
my_value = "foo bar"
print(str(my_value))
print("%s" % my_value)
print(f"{my_value}")
print(format(my_value))
print(my_value.__format__("s"))
print(my_value.__str__())
```

```python
class OpaqueClass:
    def __init__(self, x, y):
        self.x = x
        self.y = y
        
obj = OpaqueClass(1, "foo")
print(obj)
```

```python
class BetterClass:
    def __init__(self, x, y):
        self.x = x
        self.y = y
        
    def __repr__(self):
        return f"BetterClass({self.x!r}, {self.y!r})"
```

```python
obj = BetterClass(2, "foo")
print(obj)
```

```python
print(str(obj))
```

```python
class StringifiableBetterClass(BetterClass):
    def __str__(self):
        return f"({self.x}, {self.y})"
```

```python
obj2 = StringifiableBetterClass(2, "bar")
print("Human readable:", obj2)
print("Printable: ", repr(obj2))
```

## 14: Know How to Slice Sequences

The basic form of the slicing syntax is `somelist[start:end]` ,
where `start` is inclusive and `end` is exclusive:

```python
a = ["a", "b", "c", "d", "e", "f", "g", "h"]
print("Middle two: ", a[3:5])
print("All but ends:", a[1:7])
```

```python
a[5:]
```

```python
print(a[:])
print(a[:5])
print(a[:-1])
print(a[4:])
print(a[-3:])
print(a[2:5])
print(a[2:-1])
print(a[-3:-1])
```

```python
first_twenty_items = a[:20]
last_twenty_items = a[-20:]
```

```python
# a[20]
```

```python
b = a[3:]
print("Before: ", b)
b[1] = 99
print("After: ", b)
print("No change: ", a)
```

```python
print("Before ", a)
a[2:7] = [99, 22, 14]
print("After ", a)
```

```python
print("Before ", a)
a[2:3] = [47, 11]
print("After ", a)
```

```python
b = a[:] # tu otwrzysz osobna instancje, ktora zajmuje osobny blok w pamieci
b == a and b is not a
```

```python
b = a # tu tworzysz referencje, czyli jak a sie zmieni, to b tez sie zmieni
b == a and b is not a
```

```python
b = a.copy() # jak przy a[:]
b == a and b is not a
```

| Operacja               | Tworzy nową listę? | Zmiany w `a` widoczne w `b`? | Typ kopii          | Uwagi                                              |
| ---------------------- | ------------------ | ---------------------------- | ------------------ | -------------------------------------------------- |
| `b = a[:]`             | ✅ Tak              | ❌ (dla prostych typów)       | płytka kopia       | Kopiuje tylko pierwszy poziom elementów            |
| `b = a`                | ❌ Nie              | ✅ zawsze                     | ta sama referencja | `a` i `b` wskazują na ten sam obiekt               |
| `b = copy.copy(a)`     | ✅ Tak              | ❌ (dla prostych typów)       | płytka kopia       | Działa podobnie do `a[:]`                          |
| `b = copy.deepcopy(a)` | ✅ Tak              | ❌ nawet dla zagnieżdżeń      | głęboka kopia      | Kopiuje całą strukturę, w tym obiekty zagnieżdżone |

## Description:
1. Write a function that takes a string and returns every third character starting from the end, without using loops.
2. Benchmark three methods of string concatenation for a large text (e.g., 10 million characters):
   - Using the `+` operator.
   - Using `"".join()` on a list of strings.
   - Using f-strings.
3. Use unpacking instead of slicing where possible to make the code cleaner.

## Tasks:

```python
def return_e3_from_end(text: str):
    text = str(text)
    
    return text[-1::-3]

return_e3_from_end('stol z powylamywanymi nogami')
```

```python
with open('text.txt', 'r') as f:
    content = f.read()
```

```python
content_list = list(content)
```

```python
def stoper(start: float, end: float):  
    total_time = end - start
    return total_time
```

```python
def check_text(text):
    if isinstance(text, str):
        text = list(text)
    elif isinstance(text, list):
        pass
    else:
        raise TypeError('Check the input data')
```

```python
from timeit import default_timer as timer
def plus_concat(text):
    check_text(text)
    
    result = ''
    start = timer()
    for i in text:
        result += i
    end = timer()
    total_time = stoper(start, end)
    
    print(f'Concatenating with "+" last {total_time:.2f} seconds')
    
```

```python
plus_concat(content_list)
```

```python
def join_concat(text):
    check_text(text)
    
    start = timer()
    
    ''.join(text)
    
    end = timer()
    total_time = stoper(start, end)
    
    print(f'Concatenating with "join" last {total_time:.2f} seconds')
```

```python
join_concat(content_list)
```

```python
def concat_fstring(text):
    check_text(text)
    
    start = timer()
    result = ''
    for i in text:
        result = f'{result}{i}'
    
    end = timer()
    total_time = stoper(start, end)
    print(f'Concatenating with "fstring" last {total_time:.2f} seconds')
```

```python
concat_fstring(content_list)
```
