„**yields**” to po angielsku „**zwraca (kolejne wartości)** / **daje w wyniku**”.
W Pythonie widzisz to zwykle w dwóch miejscach:

1. **Słowo kluczowe `yield` (generatory)**
   `yield` zwraca wartość **i pauzuje** funkcję; przy następnym `next()`/iteracji funkcja wznawia działanie od miejsca pauzy. Funkcja z `yield` tworzy **generator** (leniwa produkcja danych, bez trzymania całej listy w pamięci).

```python
def countdown(n):
    while n > 0:
        yield n   # zwraca n i pauzuje
        n -= 1

for x in countdown(3):
    print(x)      # 3, 2, 1
```

Różnica: `return` kończy funkcję raz; `yield` może „zwracać” wiele razy podczas iteracji.

2. **Nagłówek w docstringu „Yields:”**
   W dokumentacji funkcji-generatora zamiast „Returns:” używa się „**Yields:**”, bo funkcja **produkuje kolejne elementy**.

```python
def read_chunks(fp, size=1024):
    """Read file in chunks.

    Yields:
        bytes: kolejne porcje danych o rozmiarze do `size`.
    """
    while True:
        data = fp.read(size)
        if not data:
            return          # koniec generatora
        yield data
```

Krótka ściąga:

* `yield` → zwraca wartość i wstrzymuje funkcję (generator).
* „Yields:” w docstringu → opis tego, **co generator zwraca w kolejnych krokach**.

````python
def coprime(a,b):
    for i in range(2, min(a,b)+1):
        if a % i == 0 and b % i == 0:
            return False
    return True
````

````python
assert coprime(4, 9)
assert not coprime(3, 6)
````

````python
def coprime_alternate(a, b):
    is_coprime = True
    for i in range(2, min(a, b) + 1):
        if a % i == 0 and b % i == 0:
            is_coprime = False
            break
    return is_coprime
````

## 21. Be Defensive when Iterating over Arguments

````python
def normalize(numbers):
    total = sum(numbers)
    result = []
    for value in numbers:
        percent = 100 * value / total
        result.append(percent)
    return result
````

````python
visits = [15, 35, 80]
percentages = normalize(visits)
print(percentages)
````

````python
def read_visits(data_path):
    with open(data_path) as f:
        for line in f:
            yield int(line)
````

````python
it = read_visits('my_numbers.txt')
print(list(it))
print(list(it))
````

````python
def normalize_copy(numbers):
    numbers_copy = list(numbers) # copy the iterator
    total = sum(numbers_copy)
    result = []
    for value in numbers_copy:
        percent = 100* value / total
        result.append(percent)
    return result
````

````python
it = read_visits('my_numbers.txt')
res = normalize_copy(it)
res
````

````python
def normalize_func(get_iter):
    total = sum(get_iter())
    result = []
    for value in get_iter():
        percent = 100 * value / total
        result.append(percent)
    return result
````

````python
path = 'my_numbers.txt'
percentages = normalize_func(lambda: read_visits(path))
print(percentages, sum(percentages))
````


## Funkcje all() i any() w Pythonie

### all(iterable)
- Zwraca **True**, jeśli *wszystkie* elementy w iterowalnym obiekcie są prawdziwe (truthy).  
- Jeśli iterowalny obiekt jest pusty, wynik to również **True** (logika „pusta koniunkcja jest prawdziwa”).  
- Odpowiada łańcuchowi operacji logicznych `and`.  
- Typowe zastosowanie: sprawdzanie, czy każdy element spełnia warunek.  

#### Przykład:
```python
nums = [2, 4, 6, 8]
print(all(n % 2 == 0 for n in nums))  # True, wszystkie liczby są parzyste
```
---

### any(iterable)

* Zwraca **True**, jeśli *przynajmniej jeden* element w iterowalnym obiekcie jest prawdziwy (truthy).
* Jeśli iterowalny obiekt jest pusty, wynik to **False**.
* Odpowiada łańcuchowi operacji logicznych `or`.
* Typowe zastosowanie: sprawdzanie, czy istnieje element spełniający warunek.

#### Przykład:

```python
nums = [1, 3, 7, 8, 11]
print(any(n % 2 == 0 for n in nums))  # True, bo 8 jest parzysta
```

---

### Kluczowe informacje

* Obie funkcje stosują **short-circuiting**:

  * `all()` zatrzymuje się przy pierwszym **False**.
  * `any()` zatrzymuje się przy pierwszym **True**.
* Dzięki temu są bardziej wydajne i czytelne niż ręczne pisanie pętli.
* Najczęściej używa się ich z **wyrażeniami generatorowymi** (`warunek(x) for x in iterable`), co pozwala uniknąć tworzenia tymczasowych list.

W Pythonie operator logiczny `and` nie zwraca po prostu `True` / `False`, tylko:

sprawdza wartości od lewej do prawej,

jeśli znajdzie coś `„falsy”` (np. 0, '', None, False), to od razu zwraca tę wartość i kończy,

jeśli wszystkie są `„truthy”`, to zwraca ostatnią wartość.

#### Producing Combinations of Items from Iterators

`batched`

````python
# unfortunetly I am working on python 3.11 where itertools.batched does not exist, so i have to implement function there
from itertools import islice

def batched(iterable, n):
    "Batch data into tuples of length n. The last batch may be shorter."
    # batched('ABCDEFG', 3) --> ABC DEF G
    if n < 1:
        raise ValueError('n must be at least one')
    it = iter(iterable)
    while batch := tuple(islice(it, n)):
        yield batch

it = batched([i for i in range(10)], 3)  # normally itertools.batched
print(list(it))
````

````python
import numpy as np
import math

def is_prime(n):
    if n == 1:
        return False
    
    arange = np.arange(2, math.ceil(n**(1/2))+1, 1)
    return not any((z == int(z)) for z in n/arange)
````

````python
def has_primes(nums):
    return any(is_prime(n) for n in nums)
````

````python
has_primes([1,4,6,16,23])
````

````python
from itertools import combinations, combinations_with_replacement
it = combinations([i for i in range(10)], 3)
comb = list(it)
````

````python
it = combinations_with_replacement([i for i in range(10)], 3)
comb_w_r = list(it)
````

````python
len(comb), len(comb_w_r)
````

````python
sum(comb_w_r[5])
````

````python
any(sum(i) % 15 == 0 for i in comb)
````

````python
any(sum(i) % 15 == 0 for i in comb_w_r)
````

````python
def check_atleast_one_even(nums):
    return(any(n % 2 == 0 for n in nums))
````

````python
all(check_atleast_one_even(i) for i in comb)
````

````python
all(check_atleast_one_even(i) for i in comb_w_r)
````
