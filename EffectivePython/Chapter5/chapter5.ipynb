````python
def get_median(numbers):
    count = len(numbers)
    sorted_numbers = sorted(numbers)
    middle = count // 2
    if count % 2 == 0:
        lower = sorted_numbers[middle - 1]
        upper = sorted_numbers[middle]
        median = (lower + upper) / 2
    else:
        median = sorted_numbers[middle]
    return median
        
def get_stats_more(numbers):
    minimum = min(numbers)
    maximum = max(numbers)
    count = len(numbers)
    average = sum(numbers) / count
    median = get_median(numbers)
    
    return minimum, maximum, average, median, count
````

````python
minimum, maximum, average, median, count = get_stats_more(lengths)
print(f"Min: {minimum}, Max: {maximum}")
print(f"Average: {average}, Median: {median}, Count {count}")
````

````python
from dataclasses import dataclass

@dataclass
class Stats:
    minimum: float
    maximum: float
    average: float
    median: float
    count: int
    
def get_stats_obj(numbers):
    return Stats(
        minimum = min(numbers),
        maximum = max(numbers),
        count = len(numbers),
        average= sum(numbers) / len(numbers),
        median = get_median(numbers)
    )
    
result = get_stats_obj(lengths)
print(result)
````

### Things to Remember

- Functions that return `None` to indicate special meaning are error-prone because `None` and many other values (such as zero and empty strings) evaluate to False in boolean expressions.

- Raise exceptions to indicate special situations instead of returning `None`. Expect the calling code to handle exceptions properly when they’re documented.

- Use type annotations to make it clear that a function will never return the value `None`, even in special situations.

---

### Rzeczy do zapamiętania

- Funkcje zwracające `None` w celu zaznaczenia szczególnego stanu są podatne na błędy, ponieważ `None` i wiele innych wartości (np. zero czy pusty ciąg) oceniane są jako False w wyrażeniach logicznych.

- Zamiast zwracać `None`, preferuj zgłaszanie wyjątków w sytuacjach szczególnych. Oczekuj, że kod wywołujący poprawnie obsłuży udokumentowane wyjątki.

- Używaj adnotacji typów, aby jasno wskazać, że funkcja nigdy nie będzie zwracać wartości `None`, nawet w sytuacjach specjalnych.


# 33. Know How to Closures Interact with Variable Scope and `nonlocal`

````python
def sort_priority(values, group):
    def helper(x):
        if x in group:
            return (0, x)
        return (1, x)
    
    
    values.sort(key=helper)
````

````python
numbers = [8, 3, 1, 2, 5, 4, 7, 6]
group = {2, 3, 5, 7}
sort_priority(numbers, group)
print(numbers)
````

- Python supports closures—that is, functions that refer to
variables from the scope in which they were defined. This is
why the helper function is able to access the group
argument for the sort_priority function.

- Functions are first-class objects in Python, which means
you can refer to them directly, assign them to variables, pass
them as arguments to other functions, compare them in
expressions and if statements, and so on. This is how the
sort method can accept a closure function as the key
argument.

- Python has specific rules for comparing sequences
(including tuples). It first compares items at index zero; then,
if those are equal, it compares items at index one; if they are
still equal, it compares items at index two, and so on. This is
why the return value from the helper closure causes the
sort order to have two distinct groups.

---

- Python obsługuje domknięcia (closures) — czyli funkcje, które odwołują się do zmiennych z zakresu, w którym zostały zdefiniowane. Dzięki temu funkcja pomocnicza może uzyskać dostęp do argumentu `group` w funkcji `sort_priority`.

- Funkcje są obiektami pierwszej klasy w Pythonie, co oznacza, że można się do nich odwoływać bezpośrednio, przypisywać je do zmiennych, przekazywać jako argumenty do innych funkcji, porównywać w wyrażeniach i instrukcjach warunkowych itp. Dzięki temu metoda `sort` może przyjmować funkcję domknięcia jako argument `key`.

- Python ma określone zasady porównywania sekwencji (w tym krotek). Najpierw porównuje elementy o indeksie zero; jeśli są równe, porównuje elementy o indeksie jeden; jeśli nadal równe, porównuje elementy o indeksie dwa itd. Dlatego wartość zwracana przez funkcję pomocniczą powoduje, że porządek sortowania tworzy dwie odrębne grupy.


````python
def sort_priority2(numbers, group):
    found = False
    
    def helper(x):
        if x in group:
            found = True
            return(0, x)
        return(1, x)
    
    numbers.sort(key=helper)
    return found
````

````python
found = sort_priority2(numbers, group)
print(f'Found:', found)
print(numbers)
````

When you reference a variable in an expression, the Python
interpreter traverses the nested scopes to resolve the reference
in this order:
1. The current function’s scope
2. Any enclosing scopes (such as other containing functions)
3. The scope of the module that contains the code (also called
the global scope)
4. The built-in scope (that contains functions like `len` and `str` )

---

Gdy odwołujesz się do zmiennej w wyrażeniu, interpreter Pythona przeszukuje zagnieżdżone zakresy, aby rozwiązać odwołanie, w następującej kolejności:
1. Zakres bieżącej funkcji
2. Zakresy zewnętrzne (np. inne funkcje, które zawierają daną funkcję)
3. Zakres modułu zawierającego kod (tzw. zakres globalny)
4. Zakres wbudowany (zawierający funkcje takie jak `len` i `str`)

````python
def sort_priority3(numbers, group):
    found = False
    
    def helper(x):
        nonlocal found # new
        if x in group:
            found = True
            return(0, x)
        return(1, x)
    
    numbers.sort(key=helper)
    return found
````

````python
found = sort_priority3(numbers, group)
print(f'Found:', found)
print(numbers)
````

````python
class Sorter:
    def __init__(self, group):
        self.group = group
        self.found = False
        
    def __call__(self, x):
        if x in self.group:
            self.found=True
            return(0, x)
        return(1,x)
        
````

````python
sorter = Sorter(group)
numbers.sort(key=sorter)
print('Found:', sorter.found)
print(numbers)
````

### Things to Remember

- Closure functions can refer to variables from any of the
enclosing scopes in which they were defined.
- By default, closures can’t affect enclosing scopes by
assigning variables.
- Use the nonlocal statement to indicate when a closure can
modify a variable in its enclosing scopes. Use the global
statement to do the same thing for module-level names.
- Avoid using nonlocal statements for anything beyond
simple functions.

---

### Rzeczy do zapamiętania

- Funkcje domknięcia (closures) mogą odwoływać się do zmiennych z dowolnego zakresu zewnętrznego, w którym zostały zdefiniowane.
- Domyślnie domknięcia nie mogą modyfikować zmiennych w zakresach zewnętrznych przez przypisanie.
- Użyj instrukcji nonlocal, gdy domknięcie ma modyfikować zmienną w zakresie zewnętrznym. Użyj instrukcji global, gdy chcesz to zrobić dla nazw na poziomie modułu.
- Unikaj używania nonlocal do bardziej złożonych przypadków — stosuj je tylko w prostych funkcjach.

## 34. Reduce Visual Noise with Variable Positional Arguments

````python
def log(message, values):
    if not values:
        print(message)
    else:
        values_str = ", ".join(str(x) for x in values)
        print(f'{message}: {values_str}')
        
log("My numbers are", [1, 2])
log("Hi there", [])
````

````python
def log(message, *values): # changed
    if not values:
        print(message)
    else:
        values_str = ", ".join(str(x) for x in values)
        print(f'{message}: {values_str}')
        
log("My numbers are", 1, 2)
log("Hi there") # changed
````

````python
favorites = [7, 33, 99]
log('Favorite colors', *favorites)
````

````python
def my_generator():
    for i in range(10):
        yield i

def my_func(*args):
    print(args)
    
it = my_generator()
my_func(*it)
````

````python
def log_seq(sequence, message, *values):
    if not values:
        print(f'{sequence} - {message}')
    else:
        value_str = ", ".join(str(x) for x in values)
        print(f'{sequence} - {message}: {value_str}')
        
log_seq(1, "favorites", 7, 33)
log_seq(1, "Hi there")
log_seq("Favorite numbers", 7, 33)
````

### Things to Remember

- You can have functions accept a variable number of
positional arguments by using `*args` in the `def` statement.
- You can use the items from a sequence as the positional
arguments for a function with the `*` operator.
- Using the `*` operator with a generator may cause a
program to run out of memory and crash.
- Adding new positional arguments to functions that accept
`*args` can introduce hard-to-detect bugs.

````python
def decode(data, default=None):
    """Load JSON data from a string.

    Args:
        data: JSON data to decode
        default: Value to return if decoding fails.
            Defaults to an empty dictionary.
    """
    
    try:
        return json.loads(data)
    except ValueError:
        if default is None:
            default = {}
        return default
````

````python
foo = decode("bad data")
foo["stuff"] = 5
bar = decode("also bad")
bar['meep'] = 1
print('Foo:', foo)
print('Bar:', bar)
assert foo is not bar
````

````python
def log_typed(message: str, when: datetime | None = None) -> None:
    """Log a message with a timestamp

    Args:
        message: Message to print
        when: datetime of when the message occurred.
            Defaults to the present time.
    """
    if when is None:
        when = datetime.now()
    print(f'{when}: {message}')
````

### Things to Remember
- A default argument value is evaluated only once: during
function definition at module load time. This can cause odd
behaviors for dynamic values (like function calls, newly
created objects, and container types).
- Use `None` as a placeholder default value for a keyword
argument that must have its actual default value initialized
dynamically. Document the intended default for the
argument in the function’s docstring. Check for the `None`
argument value in the function body to trigger the correct
default behavior.
- Using `None` to represent keyword argument default values
also works correctly with type annotations.

---

### Rzeczy do zapamiętania
- Wartość domyślnego argumentu jest obliczana tylko raz: podczas
definicji funkcji w czasie ładowania modułu. Może to powodować nieoczekiwane zachowania dla wartości dynamicznych (np. wywołań funkcji, nowo utworzonych obiektów i typów kontenerowych).
- Używaj `None` jako zastępczej wartości domyślnej dla argumentu nazwanego (keyword), którego rzeczywista wartość domyślna musi być zainicjalizowana dynamicznie. Udokumentuj zamierzoną wartość domyślną w docstringu funkcji. W ciele funkcji sprawdź wartość `None`, aby ustawić odpowiednie domyślne zachowanie.
- Używanie `None` jako wartości domyślnej dla argumentów nazwanych działa poprawnie również z adnotacjami typów.


## 37. Enforce Clarity with Keyword-Only and Positional Only Arguments

````python
def safe_division(
    number,
    divisor,
    ignore_overflow,
    ignore_zero_division,
):
    try:
        return number / divisor
    except OverflowError:
        if ignore_overflow:
            return 0
        else:
            raise
    except ZeroDivisionError:
        if ignore_zero_division:
            return float('inf')
        else:
            raise
````

````python
result = safe_division(1.0, 10**500, True, False)
print(result)
````

````python
result = safe_division(1.0, 0, True, True)
print(result)
````

````python
def safe_division_b(
    number,
    divisor,
    ignore_overflow = False,
    ignore_zero_division = False,
):
    try:
        return number / divisor
    except OverflowError:
        if ignore_overflow:
            return 0
        else:
            raise
    except ZeroDivisionError:
        if ignore_zero_division:
            return float('inf')
        else:
            raise
````

````python
result = safe_division_b(1.0, 10**500, ignore_overflow=True)
print(result)
result = safe_division_b(1.0, 0, ignore_zero_division=True)
print(result)
````

````python
def safe_division_d(
    numerator,
    denominator,
    *,
    ignore_overflow = False,
    ignore_zero_division = False,
):
    try:
        return numerator / denominator
    except OverflowError:
        if ignore_overflow:
            return 0
        else:
            raise
    except ZeroDivisionError:
        if ignore_zero_division:
            return float('inf')
        else:
            raise
````

````python
safe_division_d(number=2, divisor=5)
````

````python
def safe_division_e(
    numerator,
    denominator,
    /,
    *,
    ignore_overflow = False,
    ignore_zero_division = False,
):
    try:
        return numerator / denominator
    except OverflowError:
        if ignore_overflow:
            return 0
        else:
            raise
    except ZeroDivisionError:
        if ignore_zero_division:
            return float('inf')
        else:
            raise
````

````python
def safe_division_f(
    numerator,
    denominator,
    /,
    ndigits = 10,
    *,
    ignore_overflow = False,
    ignore_zero_division = False,
):
    try:
        fraction = numerator / denominator
        return round(fraction, ndigits)
    except OverflowError:
        if ignore_overflow:
            return 0
        else:
            raise
    except ZeroDivisionError:
        if ignore_zero_division:
            return float('inf')
        else:
            raise
````

````python
result = safe_division_f(22, 7)
print(result)
result = safe_division_f(22, 7, 5)
print(result)
result = safe_division_f(22, 7, ndigits=2)
print(result)
````

### Things to Remember
- Keyword-only arguments force callers to supply certain
arguments by keyword (instead of by position), which makes
the intention of a function call clearer. Keyword-only
arguments are defined after a `*` in the argument list
(whether on its own or as part of variable arguments like
`*args` ).
- Positional-only arguments ensure that callers can’t supply
certain parameters using keywords, which helps reduce
coupling. Positional-only arguments are defined before a
single `/` in the argument list.
- Parameters between the `/` and `*` characters in the
argument list may be supplied by position or keyword, which
is the default for Python parameters.

---

### Rzeczy do zapamiętania
- Argumenty tylko nazwane (keyword-only) wymuszają, aby wywołujący przekazywali niektóre argumenty przez nazwę (zamiast pozycyjnie), co sprawia, że intencja wywołania funkcji jest bardziej czytelna. Argumenty tylko nazwane definiuje się po `*` w liście argumentów (samodzielnie lub jako część zmiennej listy argumentów, np. `*args`).
- Argumenty tylko pozycyjne (positional-only) zapewniają, że wywołujący nie mogą przekazywać niektórych parametrów przez nazwę, co pomaga ograniczyć sprzężenie. Argumenty tylko pozycyjne definiuje się przed pojedynczym `/` w liście argumentów.
- Parametry umieszczone między znakami `/` i `*` w liście argumentów mogą być przekazywane zarówno pozycyjnie, jak i przez nazwę — to domyślne zachowanie parametrów w Pythonie.

## 38. Define Function Decorators with functools.wraps

````python
def trace(func):
    def wrapper(*args, **kwargs):
        args_repr = repr(args)
        kwargs_repr = repr(kwargs)
        result = func(*args, **kwargs)
        print(f'{func.__name__}'
              f'({args_repr}, {kwargs_repr})'
              f'-> {result!r}')
        return result
    return wrapper
````

````python
@trace
def fibonacci(n):
    """Return the n-th Fibonacci number

    Args:
        n (_type_): _description_
    """
    
    if n in (0, 1):
        return n
    return fibonacci(n-2) + fibonacci(n-1)
````

````python
from functools import wraps

def trace(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        args_repr = repr(args)
        kwargs_repr = repr(kwargs)
        result = func(*args, **kwargs)
        print(f'{func.__name__}'
              f'({args_repr}, {kwargs_repr})'
              f'-> {result!r}')
        return result
    return wrapper

@trace
def fibonacci(n):
    """Return the n-th Fibonacci number

    Args:
        n (_type_): _description_
    """
    
    if n in (0, 1):
        return n
    return fibonacci(n-2) + fibonacci(n-1)
````

### Things to Remember
- Decorators in Python are syntax to allow one function to
modify another function at runtime. 
- Using decorators can cause strange behaviors in tools that
do introspection, such as debuggers. 
- Use the `wraps` decorator from the `functools` built-in
module when you define your own decorators to avoid any
issues.

## 39. Prefer `functools.partial` over `lambda` Expressions for Glue Functions

````python
import math
import functools

def log_sum(log_total, value):
    log_value = math.log(value)
    return log_total + log_value

result = functools.reduce(log_sum, [10, 20, 40], 0)
print(math.exp(result))
````

````python
def log_sum_alt(value, log_total):
    log_value = math.log(value)
    return log_total + log_value
````

````python
result = functools.reduce(
    lambda total, value: log_sum_alt(value,total),
    [10,20,40],
    0,
)
result
````

````python
def log_sum_for_reduce(total, value):
    return log_sum_alt(value, total)
````

````python
def logn_sum(base, logn_total, value):
    logn_value = math.log(value, base)
    return logn_total + logn_value
````

````python
result = functools.reduce(
    lambda total, value: logn_sum(10, total, value),
    [10, 20, 40],
    0,
)

print(math.pow(10, result))
````

````python
result = functools.reduce(
    functools.partial(logn_sum, 10),
    [10, 20, 40],
    0,
)
print(math.pow(10, result))
````

````python
def logn_sum_last(logn_total, value, *, base = 10):
    logn_value = math.log(value, base)
    return logn_total + logn_value
````

````python
log_sum_e = functools.partial(
    logn_sum_last,
    base = math.e
)

print(log_sum_e(3, math.e**10))
````

````python
print(log_sum_e.args, log_sum_e.keywords, log_sum_e.func)
````

### Things to Remember
- `lambda` expressions can succinctly make two function
interfaces compatible by reordering arguments or pinning
certain parameter values.
- The `partial` function from the `functools` built-in is a
general tool for creating functions with pinned positional and
keyword arguments.
- Use `lambda` instead of `partial` if you need to reorder the
arguments of a wrapped function.

---

### Rzeczy do zapamiętania
- Wyrażenia `lambda` mogą zwięźle sprawić, że interfejsy dwóch funkcji będą zgodne poprzez przetasowanie kolejności argumentów lub „przypięcie” określonych wartości parametrów.
- Funkcja `partial` z wbudowanego modułu `functools` to ogólne narzędzie do tworzenia funkcji z przypiętymi argumentami pozycyjnymi i nazwanymi (keyword).
- Użyj `lambda` zamiast `partial`, jeśli musisz zmienić kolejność argumentów opakowywanej funkcji.
