**Note** Tensor datatypes is one of the 3 big errors you'll run into with PyTorch & DL
1. Tensors not right datatype
2. Tensors not right shape
3. Tensors not on the right device

````python
#float32
float_32_tensor = torch.tensor([3.0, 6.0, 9.0],
                                dtype=None,
                                device=None,
                                requires_grad=False)
float_32_tensor.dtype
````

````python
float_16_tensor = float_32_tensor.type(torch.float16)
float_16_tensor.dtype
````

1. Tensors not right datatype - ``` tensor.dtype ```
2. Tensors not right shape - ``` tensor.shape ```
3. Tensors not on the right device - ```tensor.device ```

````python
#reshaping, viewing, stacking and squeezing tensors
#squeeze - removes all dimensions of size 1 from a tensor
#unsqueeze - adds a dimension of size 1 to a tensor
#permute - return a view of the tensor with dimensions permuted in a specified order
````

````python
y = torch.arange(1., 10.)
y,y.shape
````

````python
y_reshaped = y.reshape(1, 9)
y_reshaped
````

````python
y_reshaped.reshape(9,1)
````

````python
y.reshape(3,3)
````

````python
y_reshaped.reshape(-1, 1)
````

````python
z = y.view(3,3)
z
````

````python
y[2] = 100
z
````

````python
z[1] = 200
y
````

````python
y_stacked = torch.stack([y, y, y], dim=1)
y_stacked
````

````python
y_reshaped.squeeze()
````

````python
print(f'before: {y_reshaped.shape}')
print(f'after: {y_reshaped.squeeze().shape}')
````

````python
y_squeezed = y_reshaped.squeeze()
````

````python
# torch.unsqueze() - adds a dimension of size 1 to a tensor
print(f'before: {y_squeezed.shape}')
print(f'after: {y_squeezed.unsqueeze(dim=1).shape}')
````

````python
x_original = torch.rand(size=(224, 224, 3))
````

````python
x_permuted = x_original.permute(1, 2, 0) # shifts axis 0->1, 1->2, 2->0
````

````python
print(f'Previous shape: {x_original.shape}')
print(f'Permuted shape: {x_permuted.shape}')
````

````python
x_original[0, 0, 0] = 100
x_original[0, 0, 0], x_permuted[0, 0, 0] # should be the same value in both tensors
````

````python
## reproducibility (trying to take random out of random)


torch.manual_seed(42) # works only for one next using
random_tensor_A = torch.rand(3, 4)
random_tensor_B = torch.rand(3, 4)

print(random_tensor_A == random_tensor_B) 
````
